```python
df["Year"]=df["order_date"].dt.year
df["Month"]=df["order_date"].dt.month
df["Month_Name"]=df["order_date"].dt.month_name()
```

```python
df[["order_date","Year","Month","Month_Name"]].head(10)
```

```python
df["calculated_revenue"] = (
    df["quantity"] * df["unit_price"] * (1 - df["discount"])
)
```

```python
df[["quantity", "unit_price", "discount", "revenue", "calculated_revenue"]].head(10)
```

```python
df["revenue_difference"] = (
    df["revenue"] - df["calculated_revenue"]
)
df["revenue_difference"].abs().max()
```

```python
df["calculated_revenue"] = df["calculated_revenue"].round(2)
```

```python
df["revenue_difference"] = (
    df["revenue"] - df["calculated_revenue"]
)
df["revenue_difference"].abs().max()
```

```python
df.drop(columns=["calculated_revenue", "revenue_difference"], inplace=True)
```

```python
#which product category generates the most revenue
category_sales = df.groupby("product_category")["revenue"].sum()
category_sales.sort_values(ascending=False)
```

```python
#calculate the no.of orders by category
orders_by_category = df.groupby("product_category")["order_id"].count()
orders_by_category.sort_values(ascending=False)
```

```python
# Average revenue by category
avg_revenue_category = df.groupby("product_category")["revenue"].mean()
avg_revenue_category.sort_values(ascending=False)
```

```python
# calculate revenue contribution
category_revenue = df.groupby("product_category")["revenue"].sum()
category_percentage = (category_revenue / df["revenue"].sum()) * 100
category_percentage.sort_values(ascending=False)
```

```python
# Monthly sales analysis
# How much revenue did we generate in each month
monthly_sales = df.groupby(["Year", "Month", "Month_Name"])["revenue"].sum().reset_index()
monthly_sales = monthly_sales.sort_values(["Year", "Month"])
monthly_sales
monthly_sales
```

```python
# Highest revenue month
monthly_sales.loc[monthly_sales["revenue"].idxmax()]
```

```python
# Lowest revenue month
monthly_sales.loc[monthly_sales["revenue"].idxmin()]
```

```python
# Monthly avg revenue
average_monthly_revenue = monthly_sales["revenue"].mean()
round(average_monthly_revenue, 2)
```

```python
# Count orders in each month
monthly_orders = df.groupby(["Year", "Month", "Month_Name"])["order_id"].count().reset_index()
monthly_orders
```

```python
# Avg order value by month
monthly_aov = df.groupby(["Year", "Month", "Month_Name"])["revenue"].mean().reset_index()
monthly_aov = monthly_aov.rename(columns={"revenue": "average_order_value"})
monthly_aov
```

```python
# Monthly revenue growth
monthly_sales["revenue_growth_%"] = monthly_sales["revenue"].pct_change() * 100
monthly_sales
```

```python
# Best month by no.of orders
monthly_orders.loc[monthly_orders["order_id"].idxmax()]
```

```python
# Month with highest avg
monthly_aov.loc[monthly_aov["average_order_value"].idxmax()]
```

```python
# Month with highest revenue growth
monthly_sales.loc[monthly_sales["revenue_growth_%"].idxmax()]
```

```python
# Month with biggest revenue decline
monthly_sales.loc[monthly_sales["revenue_growth_%"].idxmin()]
```

```python
# Regional sales analysis
# calculate revenue by region and sort
region_revenue = df.groupby("region")["revenue"].sum().reset_index()
region_revenue = region_revenue.sort_values("revenue", ascending=False)
region_revenue
```

```python
# calculate orders by region
region_orders = df.groupby("region")["order_id"].count().reset_index()
region_orders = region_orders.sort_values("order_id", ascending=False)
region_orders
```

```python
# Avg revenue per order by region
region_aov = df.groupby("region")["revenue"].mean().reset_index()
region_aov = region_aov.rename(columns={"revenue": "average_order_value"})
region_aov = region_aov.sort_values("average_order_value", ascending=False)
region_aov
```

```python
# revenue contribution by region
region_revenue["revenue_percentage"] = (
    region_revenue["revenue"] / region_revenue["revenue"].sum()
) * 100
region_revenue["revenue_percentage"] = region_revenue["revenue_percentage"].round(2)
region_revenue
```

```python
# Region with highest revenue
region_revenue.loc[region_revenue["revenue"].idxmax()]
```

```python
# Region with highest no.of orders
region_orders.loc[region_orders["order_id"].idxmax()]
```

```python
# Region with highest avg
region_aov.loc[region_aov["average_order_value"].idxmax()]
```

```python
# Product Analysis
# Revenue by product category
category_quantity = df.groupby("product_category")["quantity"].mean().reset_index()
category_quantity = category_quantity.sort_values("quantity", ascending=False)
category_quantity
```

```python
# Total quantity slod by category
category_total_quantity = df.groupby("product_category")["quantity"].sum().reset_index()
category_total_quantity = category_total_quantity.sort_values(
    "quantity", ascending=False
)
category_total_quantity
```

```python
# Revenue by payment method
payment_revenue = df.groupby("payment_method")["revenue"].sum().reset_index()
payment_revenue = payment_revenue.sort_values("revenue", ascending=False)
payment_revenue
```

```python
# orders by payment method
payment_orders=df.groupby("payment_method")["order_id"].count().reset_index()
payment_orders=payment_orders.sort_values("order_id", ascending=False)
payment_orders
```

```python
# Customer Analysis
# Revenue by customer
customer_revenue = df.groupby("customer_id")["revenue"].sum().reset_index()
customer_revenue = customer_revenue.sort_values("revenue", ascending=False)
customer_revenue.head(10)
```

```python
# No.of orders per customer
customer_orders = df.groupby("customer_id")["order_id"].count().reset_index()
customer_orders = customer_orders.sort_values("order_id", ascending=False)
customer_orders.head(10)
```

```python
# Identify repeat customers
repeat_customers = customer_orders[customer_orders["order_id"] > 1]
repeat_customers
```

```python
# Average customer rating
average_rating = df["customer_rating"].mean()
round(average_rating, 2)
```
